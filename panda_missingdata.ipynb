```python
import numpy as np
import pandas as pd
```

```python
data={
    'A':[1,2,np.nan,4,5],
    'B':[5,np.nan,7,np.nan,6],
    'C':[np.nan,np.nan,11,12,np.nan],
    'D':[13,14,np.nan,np.nan,17]
}
df=pd.DataFrame(data)
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

Removing the data

```python
df.dropna(thresh=3)
```

Filling the missing data

```python
df.fillna(0)
```

```python
values={'A':11,'B':22,'C':33,'D':44}
df.fillna(value=values)
```

```python
df.fillna(df.mean)
```
