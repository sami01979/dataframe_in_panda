```python
import numpy as np
import pandas as pd
```

```python
employees=pd.DataFrame({
    'employee_id':[1,2,3,4],
    'name':['Bob','Jake','Lisa','Sue'],
    'group':['Accounting','Engineering','Engineering','HR']
})
salaries=pd.DataFrame({
    'employee_id':[1,2,3,6,5,4],
    'salary':[70000,80000,120000,90000,56000,23000],
    'bonus':[1000,2000,3000,4000,2000,9000]

})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries)
```

```python
pd.merge(employees,salaries,on='employee_id',how='outer')
```

Concatenation

```python
df1=pd.DataFrame({
    'A':['A0','A1','A2','A3'],
    'B':['B0','B1','B2','B3'],
    'C':['C0','C1','C2','C3']
})
df2=pd.DataFrame({
    'A':['A4','A5','A6','A7'],
    'B':['B4','B5','B6','B7'],
    'C':['C4','C5','C6','C7']
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],axis=1)
```

Joining

```python
df1=pd.DataFrame({
    'name':['Bob','Jake','Lisa','Sue']

},index=[1,2,3,4])
df2=pd.DataFrame({
    'score':[87,98,45,53]
},index=[2,3,4,5])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2,how='inner')
```

```python
df1.join(df2,how='outer')
```
